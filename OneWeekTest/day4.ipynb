Create a function that returns the factorial of a number.

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
num = int(input("enter a number"))
if (num == 0):
    print("Enter a value greater than zero ", num)
else:
    print(f"The factorial of the number is {num} =", factorial(num))
```

Write a function that counts the frequency of each character in a string.

```python
def char_frequency(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

s1 = input("enter a string ")
print("typed string  ",s1,"\n")
print(char_frequency(s1))
```

Find Common Elements in Multiple Sets.
s1 = {1, 2, 3, 4}
s2 = {2, 3, 5}
s3 = {0, 2, 3, 6}

```python


def common(s1,s2,s3):
    # l = [a for a in s1 for b in s2 if a and b in s3  ]
    l = []
    print(type(l))
    for a in s1:
        for b in s2:
            if (a == b):
                if(a in s3):
                    l.append(a)
                    break
    return l

# a = set()

s1 = {1, 2, 3, 4}
s2 = {2, 3, 5}
s3 = {0, 2, 3, 6}
print("-------"*20,"\n","Common Elements\n")
print(set(common(s1,s2,s3)))           #  logic 1
print(set.intersection(s1,s2,s3))      #  logic 2
```

Find elements that are in A or B but not in both
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

```python
def notInBoth(s1,s2):
    lcommon = []
    # print(type(l))
    for a in s1:
        if(a in s2):
            lcommon.append(a)
            
        
    # print(lcommon)
    lnotcommon = []
    for a in s1:
        if a not in lcommon:
            lnotcommon.append(a)
    for a in s2:
        if a not in lcommon:
            lnotcommon.append(a)
            
    return set(lnotcommon)
    

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
C = {1,2,3}

print("--"*20,"\n","Not common in both A or B")
print(notInBoth(A,B)) # logic 1

D= notInBoth(A,B)
print(notInBoth(C,D))

print(A.symmetric_difference(B))  # logic 2
```

Write a function to return all unique alphabet characters (case-insensitive)
        sentence = "Python is powerful and fun!"

```python
def unique_alphabets(sentence):
    sentence = sentence.lower() 
    unique_chars = set()

    for char in sentence:
        if char.isalpha(): 
            unique_chars.add(char)

    return sorted(unique_chars)


sentence = "Python is powerful and fun!"
print(unique_alphabets(sentence))
```

Reverse Key-Value Uniquely
        d = {'a': 1, 'b': 2, 'c': 1}
        Expected answer like this: {1: ['a', 'c'], 2: ['b']}

```python
def reverse1(d):
    neew = {}
    # d.inverse()
    for key, value in d.items():
        neew.setdefault(value,[]).append(key)
    print(neew)
    
d = {'a': 1 , 'b': 2, 'c': 1}
reverse1(d)
```
