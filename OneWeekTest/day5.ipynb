Q3. Number Decomposition
Given an integer n, write a program to return all unique combinations of positive integers that sum up to n.


```python

n = int(input("enter  anumber"))

l = []

for i in range(0,n+1):
    res = []
    for j in range(i+1,n+1):
        if(j>n):
            break
        res.append(j) 
        if(sum(res) == n):
            l.append(res)
            res.clear()
            
print(l)
```

Q5. Set Operations Challenge
You are given two lists:
list1 = [1, 2, 3, 4, 5, 6]

```python
list1 = [1, 2, 3, 4, 5, 6]
list2 = [4, 5, 6, 7, 8, 9]

set1 = set(list1)
set2 = set(list2)

print("set intersection", set1.intersection(set2))
print("set union", set1.union(set2))
print("set symmetric_differnce or XOR", set1.symmetric_difference(set2)) 
```
